```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employee_data (2).csv")
print(df.head())

# Display employee data
print("Employee Data:")
print(df)

# Calculate average salary department-wise
avg_salary = df.groupby("Department")["Salary"].mean()

print("\nAverage Salary Department-wise:")
print(avg_salary)

# Draw bar graph
plt.figure(figsize=(8,5))
avg_salary.plot(kind="bar", color="skyblue")

plt.title("Average Salary Department-wise")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```
